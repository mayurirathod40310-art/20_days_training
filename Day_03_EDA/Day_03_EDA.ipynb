```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("students.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
department_counts = df['Department'].value_counts()

print(department_counts)

department_counts.plot(kind='bar', title='Number of Students in Each Department')

plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.hist(df['Marks'].dropna(), bins=5, edgecolor='black')

plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.hist(df['Age'].dropna(), bins=5, edgecolor='black')

plt.title('Distribution of Student Ages')
plt.xlabel('Age')
plt.ylabel('Number of Students')

plt.show()
```

```python
correlation = df[['Age', 'Marks']].corr()

plt.figure(figsize=(6, 4))

plt.imshow(correlation, cmap='coolwarm', vmin=-1, vmax=1)

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(j, i, round(correlation.iloc[i, j], 2),
                 ha='center', va='center')

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(7, 4))

df.boxplot(column='Marks', by='Department')

plt.title('Marks Comparison Across Departments')
plt.suptitle('')
plt.xlabel('Department')
plt.ylabel('Marks')

plt.show()
```

Observations :

Department Distribution: The bar chart shows the number of students in each department.



Marks Distribution: The histogram shows that student marks range from 65 to 95.



Age Distribution: The age histogram shows that student ages range from 20 to 23 years.



Correlation: The heatmap shows the relationship between age and marks. Correlation does not necessarily imply causation.



Department-wise Performance: The box plot helps compare the median and spread of marks across departments.
