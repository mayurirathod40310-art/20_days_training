```python
import numpy as np
import pandas as pd

print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)

print("Day 2 - NumPy and Pandas Started!")
```

```python
# Creating a NumPy array
marks = np.array([85, 90, 78, 92, 88])

print("Student Marks:", marks)
print("Average Marks:", np.mean(marks))
print("Maximum Marks:", np.max(marks))
print("Minimum Marks:", np.min(marks))
```

```python
# Creating a Pandas Series
student_names = pd.Series(["Amit", "Riya", "Neha", "Rahul"])

print(student_names)
```

```python
# Creating a Pandas DataFrame
data = {
    "Name": ["Amit", "Riya", "Neha"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 106, 108],
    "Name": ["Amit", "Riya", "Neha", "Rahul", "Sneha", "Karan", "Karan", "Priya"],
    "Age": [21, 22, 20, None, 21, 23, 23, 22],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE", "CSE", "ECE"],
    "Marks": [85, 90, 78, 65, None, 88, 88, 95]
}

df = pd.DataFrame(data)

display(df)
```

```python
print("First 5 Records:")
display(df.head())

print("Dataset Shape:", df.shape)

print("Dataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate Records:")
print(df.duplicated().sum())

print("\nDuplicate Rows:")
display(df[df.duplicated()])

print("Duplicate Student IDs:")
print(df["Student_ID"].duplicated().sum())
```

```python
print("Data Types:")
print(df.dtypes)

print("\nStatistical Summary:")
display(df.describe())
```

```python
df_cleaned = df.drop_duplicates().copy()

df_cleaned["Age"] = df_cleaned["Age"].fillna(df_cleaned["Age"].mean())

df_cleaned["Marks"] = df_cleaned["Marks"].fillna(df_cleaned["Marks"].mean())

print("Cleaned Dataset:")
display(df_cleaned)
```

```python
high_scorers = df_cleaned[df_cleaned["Marks"] > 80]

print("Students Scoring Above 80:")
display(high_scorers)
```

```python
sorted_students = df_cleaned.sort_values(by="Marks", ascending=False)

print("Students Sorted by Marks:")
display(sorted_students)
```

```python
print("Final Cleaned Dataset:")
display(df_cleaned)

print("Final Dataset Shape:", df_cleaned.shape)

print("Missing Values:", df_cleaned.isnull().sum().sum())

print("Duplicate Rows:", df_cleaned.duplicated().sum())
```

```python
department_summary = df_cleaned.groupby("Department")["Marks"].mean()

print("Average Marks Department-wise:")
print(department_summary)
```

```python
department_report = df_cleaned.groupby("Department").agg(
    Total_Students=("Name", "count"),
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max")
)

print("Department-wise Report:")
display(department_report)
```

```python
department_info = pd.DataFrame({
    "Department": ["CSE", "IT", "ECE"],
    "Location": ["Block A", "Block B", "Block C"]
})

print("Department Information:")
display(department_info)
```

```python
merged_data = pd.merge(
    df_cleaned,
    department_info,
    on="Department",
    how="left"
)

print("Merged Student Data:")
display(merged_data)
```

```python
df_cleaned.to_csv("cleaned_students.csv", index=False)

department_report.to_csv("department_summary.csv")

print("Files exported successfully!")
```

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Total Records (Before Cleaning)",
        "Total Records (After Cleaning)",
        "Total Columns",
        "Missing Values (Before Cleaning)",
        "Missing Values (After Cleaning)",
        "Duplicate Rows (Before Cleaning)",
        "Duplicate Rows (After Cleaning)",
        "Duplicate Student IDs (After Cleaning)"
    ],

    "Result": [
        len(df),
        len(df_cleaned),
        len(df_cleaned.columns),
        df.isnull().sum().sum(),
        df_cleaned.isnull().sum().sum(),
        df.duplicated().sum(),
        df_cleaned.duplicated().sum(),
        df_cleaned["Student_ID"].duplicated().sum()
    ]
})

print("DATA QUALITY REPORT")
display(quality_report)
```

```python
quality_report.to_csv("data_quality_report.csv", index=False)

print("Data Quality Report exported successfully!")
```
